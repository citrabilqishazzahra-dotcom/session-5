```python

print("===== MENU RESTORAN =====")
print("1. Nasi Goreng (Rp15.000)")
print("2. Mie Ayam (Rp12.000)")
print("3. Es Teh (Rp3.000)")

pilihan = int(input("Masukkan pilihan (1-3): "))
jumlah = int(input("Masukkan jumlah pesanan: "))

# Menentukan harga menggunakan match-case
match pilihan:
    case 1:
        harga = 15000
    case 2:
        harga = 12000
    case 3:
        harga = 3000
    case _:
        print("Menu tidak ada")
        harga = 0

# Menghitung total pembayaran
if harga > 0:
    total = harga * jumlah
    print("Total bayar: Rp", total)
```
