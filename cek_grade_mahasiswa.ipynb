```python
nilai = int(input("Masukkan nilai mahasiswa: "))

if nilai >= 85 and nilai <= 100:
    grade = "A"
elif nilai >= 70:
    grade = "B"
elif nilai >= 55:
    grade = "C"
elif nilai >= 0:
    grade = "D (Remedial)"
else:
    grade = "Nilai tidak valid"

print(f"Nilai {nilai} = Grade {grade}")
```
